```python
# define numerical & categorical columns
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

# print columns
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
# Feature Engeneering

# Avoid division by zero
df['Income_to_Experience_Ratio'] = df['MonthlyIncome'] / (df['TotalWorkingYears'] + 1)

df['Promotion_Delay_Ratio'] = df['YearsSinceLastPromotion'] / (df['YearsAtCompany'] + 1)

df['Engagement_Score'] = (
    df['JobSatisfaction'] +
    df['WorkLifeBalance'] +
    df['EnvironmentSatisfaction'] +
    df['RelationshipSatisfaction']
) / 4

df['Workload_Stress_Flag'] = np.where(
    (df['OverTime'] == 'Yes') & (df['WorkLifeBalance'] <= 2),
    1, 0
)

df['HighPerf_NoPromo'] = np.where(
    (df['PerformanceRating'] >= 4) & (df['YearsSinceLastPromotion'] > 3),
    1, 0
)
```
